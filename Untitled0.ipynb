```python
import turtle
import math

def heart(t):
    # Equações paramétricas corretas para o coração
    x = 16 * (math.sin(t) ** 3)
    y = 13 * math.cos(t) - 5 * math.cos(2 * t) - 2 * math.cos(3 * t) - math.cos(4 * t)
    return x, y

screen = turtle.Screen()
screen.setup(600, 600)
screen.bgcolor("pink")
screen.colormode(1.0)

t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.width(2)

N, K = 140, 2.1
SCALE = 15
bright = (0.95, 0.15, 0.1)
dark = (0.3, 0.0, 0.0)

for i in range(N):
    ang = 2 * math.pi * i / N
    hx, hy = heart(ang)
    hx, hy = hx * SCALE, hy * SCALE
    ex, ey = hx * K, hy * K
    steps = 10

    t.penup()
    t.goto(hx, hy)
    t.pendown()

    for j in range(steps):
        f = (j + 1) / steps
        # Interpolação simples de cores entre o claro e o escuro
        col = (
            bright[0] + (dark[0] - bright[0]) * f,
            bright[1] + (dark[1] - bright[1]) * f,
            bright[2] + (dark[2] - bright[2]) * f
        )
        t.pencolor(col)
        t.goto(hx + (ex - hx) * f, hy + (ey - hy) * f)

turtle.done()
```
